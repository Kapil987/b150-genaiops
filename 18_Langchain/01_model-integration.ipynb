### Streaming And Batch

### Streaming

Most AI models can stream their responses while generating them.Instead of waiting for the complete answer, the output appears gradually in real time.
This makes the application feel faster and improves the overall user experience, especially for long responses. 

When you call `stream()`, it returns an iterator that provides small chunks of the response as they are generated.
You can loop through these chunks and display the response live to the user.

````python
llm.invoke("Write me a 200 words paragraph on cow")
````

````python
for chunk in llm.stream("Write me a 200 words paragraph on cow"):
    print(chunk.text, end="|", flush=True)
````

````python
llm.invoke("Does bats lay eggs?")
````

````python
for chunk in llm.stream("Does bats lay eggs?"):
    print(chunk.text, end="", flush=True)
````


#### Batch
Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:
https://reference.langchain.com/python/langchain-core/runnables/base/Runnable/batch

````python
responses = llm.batch([
    "who landid on moon first?",
    "is tesla an electric car?",
    "What is super computer?"
])
for response in responses:
    print(response)
````

````python
llm.batch(
    ["who landid on moon first?",
    "is tesla an electric car?",
    "What is super computer?"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
````
