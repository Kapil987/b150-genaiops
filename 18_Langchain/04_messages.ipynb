#### Messages

Messages are the basic way LangChain manages communication with LLMs. They store the conversation history and help the model understand the context of an interaction.

Each message usually contains:

* **Role** – Defines who sent the message, such as system, user, or assistant.
* **Content** – The actual information being shared, like text, images, audio, or files.
* **Metadata** – Extra details such as token usage, message IDs, timestamps, or response-related information.

LangChain uses a common message format across different AI providers, making it easier to switch between models while keeping the behavior consistent.

### Text Prompts
Text prompts are strings - ideal for straightforward generation tasks where you don’t need to retain conversation history.

```python
model.invoke("what is system prompt?")
```

Use text prompts when:
- You have a single, standalone request
- You don’t need conversation history
- You want minimal code complexity

### Message Prompts
Alternatively, you can pass in a list of messages to the model by providing a list of message objects.


Message types
- System message - Tells the model how to behave and provide context for interactions
- Human message - Represents user input and interactions with the model
- AI message - Responses generated by the model, including text content, tool calls, and metadata
- Tool message - Represents the outputs of tool calls

### System Message

A SystemMessage gives the model a starting set of instructions that guides how it should behave. It helps define the assistant’s role, response style, tone, and any rules it should follow during the conversation.
example: medical expert, insurance expert etc.

### Human Message

A `HumanMessage` is used to represent input coming from the user. It can include text, images, audio, documents, or other types of multimodal content during the conversation.

### AI Message

An `AIMessage` stores the response generated by an AI model after you send it a request. It can contain normal text replies, tool call details, images or other multimodal content, along with extra metadata provided by the model provider that you can use later.


### Tool Message
For models that support tool calling, AI messages can contain tool calls. Tool messages are used to pass the results of a single tool execution back to the model.

```python
# Testing of simulating Tools call without real tool call
from langchain.messages import AIMessage
from langchain.messages import ToolMessage

# After a model makes a tool call
# (Here, we demonstrate manually creating the messages )
ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {"location": "San Francisco"},
        "id": "call_104"
    }]
)

# Execute tool and create result message
weather_result = "Sunny, 65°F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_104"  # Must match the call ID
)

# Continue conversation
messages = [
    HumanMessage("What's the weather in San Francisco?"),
    ai_message,  # Model's tool call
    tool_message,  # Tool execution result
]
response = model.invoke(messages)  # Model processes the result
```
