```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# -----------------------------
# DOCUMENTS
# -----------------------------

documents = [
    "Azure AI Foundry supports deployment automation using GitHub Actions workflows.",

    "LangChain is commonly used for building Retrieval-Augmented Generation applications.",

    "LangGraph helps create stateful AI agents with multi-step reasoning capabilities.",

    "Kubernetes automates container deployment, scaling, and management in cloud environments.",

    "GitHub Actions enables CI/CD pipelines directly inside GitHub repositories.",

    "Terraform is widely used for Infrastructure as Code across AWS and Azure platforms.",

    "FAISS provides efficient vector similarity search for embedding-based retrieval systems.",

    "Sentence Transformers convert text into embeddings for semantic search and RAG systems.",

    "Azure DevOps supports automated build and release pipelines for enterprise applications.",

    "Large Language Models can use retrieved context to reduce hallucinations in AI responses."
]

```

```python
# -----------------------------
# LOAD EMBEDDING MODEL
# -----------------------------

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# -----------------------------
# CREATE EMBEDDINGS
# -----------------------------

doc_embeddings = model.encode(documents)

doc_embeddings = np.array(
    doc_embeddings
).astype("float32")

# -----------------------------
# CREATE FAISS INDEX
# -----------------------------

dimension = doc_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(doc_embeddings)
```

```python

# -----------------------------
# USER QUERY
# -----------------------------

# query = "How can GitHub Actions help deployment?"
query = "Explain FAISS?"

query_embedding = model.encode([query])

query_embedding = np.array(
    query_embedding
).astype("float32")

# -----------------------------
# SEARCH
# -----------------------------

k = 2

distances, indices = index.search(
    query_embedding,
    k
)

# -----------------------------
# RESULTS
# -----------------------------

print("\nTop Results:\n")

for idx in indices[0]:
    print(documents[idx])
    print()
```
